```python
# ============================================================
# APEXPLANET DATA ANALYTICS INTERNSHIP
# TASK 2 - EXPLORATORY DATA ANALYSIS (EDA)
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os

# ------------------------------------------------------------
# 1. LOAD DATASET
# ------------------------------------------------------------

file_path = "ApexPlanet_DataAnalytics_Dataset.xlsx"

df = pd.read_excel(file_path)

print("=" * 70)
print("APEXPLANET - TASK 2")
print("EXPLORATORY DATA ANALYSIS & BUSINESS INTELLIGENCE")
print("=" * 70)

print("\nDataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

display(df.head())


# ------------------------------------------------------------
# 2. BASIC INFORMATION
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("DATASET INFORMATION")
print("=" * 70)

print("\nShape:")
print(df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())


# ------------------------------------------------------------
# 3. CLEAN DATA
# ------------------------------------------------------------

# Convert date
df["Order_Date"] = pd.to_datetime(
    df["Order_Date"],
    errors="coerce"
)

# Numerical columns
numeric_columns = [
    "Age",
    "Quantity",
    "Unit_Price",
    "Total_Sales"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

# Repair sequential Order IDs
expected_ids = [
    "ORD" + str(100002 + i)
    for i in range(len(df))
]

df["Expected_Order_ID"] = expected_ids

df.loc[
    df["Order_ID"] != df["Expected_Order_ID"],
    "Order_ID"
] = df.loc[
    df["Order_ID"] != df["Expected_Order_ID"],
    "Expected_Order_ID"
]

df.drop(
    columns=["Expected_Order_ID"],
    inplace=True
)

# Missing values
df["Age"] = df["Age"].fillna(
    df["Age"].median()
)

df["City"] = df["City"].fillna(
    "Unknown"
)

# Standardize text
text_columns = [
    "Gender",
    "City",
    "Product",
    "Category"
]

for col in text_columns:
    df[col] = df[col].astype(str).str.strip().str.title()


# ------------------------------------------------------------
# 4. FEATURE ENGINEERING
# ------------------------------------------------------------

df["Year"] = df["Order_Date"].dt.year

df["Month"] = df["Order_Date"].dt.month

df["Month_Name"] = df["Order_Date"].dt.month_name()

df["Quarter"] = (
    "Q" +
    df["Order_Date"].dt.quarter.astype(str)
)

df["Age_Group"] = pd.cut(
    df["Age"],
    bins=[17, 25, 35, 50, 65],
    labels=[
        "18-25",
        "26-35",
        "36-50",
        "51-65"
    ]
)


# ------------------------------------------------------------
# 5. DESCRIPTIVE STATISTICS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("DESCRIPTIVE STATISTICS")
print("=" * 70)

display(
    df[
        [
            "Age",
            "Quantity",
            "Unit_Price",
            "Total_Sales"
        ]
    ].describe()
)


# ------------------------------------------------------------
# 6. KEY BUSINESS METRICS
# ------------------------------------------------------------

total_sales = df["Total_Sales"].sum()

average_sales = df["Total_Sales"].mean()

minimum_sales = df["Total_Sales"].min()

maximum_sales = df["Total_Sales"].max()

total_quantity = df["Quantity"].sum()

total_orders = df["Order_ID"].nunique()

unique_customers = df["Customer_ID"].nunique()

average_age = df["Age"].mean()


print("\n" + "=" * 70)
print("KEY BUSINESS METRICS")
print("=" * 70)

print(f"\nTotal Sales       : ₹{total_sales:,.2f}")
print(f"Total Orders      : {total_orders:,}")
print(f"Unique Customers  : {unique_customers:,}")
print(f"Total Quantity    : {total_quantity:,}")
print(f"Average Sales     : ₹{average_sales:,.2f}")
print(f"Minimum Sale      : ₹{minimum_sales:,.2f}")
print(f"Maximum Sale      : ₹{maximum_sales:,.2f}")
print(f"Average Age       : {average_age:.2f}")


# ------------------------------------------------------------
# 7. SALES BY PRODUCT
# ------------------------------------------------------------

product_sales = (
    df.groupby("Product")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("\n" + "=" * 70)
print("SALES BY PRODUCT")
print("=" * 70)

display(product_sales)


plt.figure(figsize=(10, 6))

product_sales.plot(
    kind="bar"
)

plt.title("Total Sales by Product")

plt.xlabel("Product")

plt.ylabel("Total Sales")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 8. SALES BY CATEGORY
# ------------------------------------------------------------

category_sales = (
    df.groupby("Category")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("\n" + "=" * 70)
print("SALES BY CATEGORY")
print("=" * 70)

display(category_sales)


plt.figure(figsize=(10, 6))

category_sales.plot(
    kind="bar"
)

plt.title("Total Sales by Category")

plt.xlabel("Category")

plt.ylabel("Total Sales")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 9. SALES BY CITY
# ------------------------------------------------------------

city_sales = (
    df.groupby("City")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("\n" + "=" * 70)
print("SALES BY CITY")
print("=" * 70)

display(city_sales)


plt.figure(figsize=(10, 6))

city_sales.plot(
    kind="bar"
)

plt.title("Total Sales by City")

plt.xlabel("City")

plt.ylabel("Total Sales")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 10. SALES BY GENDER
# ------------------------------------------------------------

gender_sales = (
    df.groupby("Gender")["Total_Sales"]
    .sum()
)

print("\n" + "=" * 70)
print("SALES BY GENDER")
print("=" * 70)

display(gender_sales)


plt.figure(figsize=(7, 5))

gender_sales.plot(
    kind="bar"
)

plt.title("Total Sales by Gender")

plt.xlabel("Gender")

plt.ylabel("Total Sales")

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 11. CUSTOMER AGE DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

plt.hist(
    df["Age"],
    bins=10
)

plt.title("Customer Age Distribution")

plt.xlabel("Age")

plt.ylabel("Number of Customers")

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 12. SALES DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

plt.hist(
    df["Total_Sales"],
    bins=30
)

plt.title("Distribution of Total Sales")

plt.xlabel("Total Sales")

plt.ylabel("Frequency")

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 13. QUANTITY DISTRIBUTION
# ------------------------------------------------------------

quantity_distribution = (
    df["Quantity"]
    .value_counts()
    .sort_index()
)

print("\nQuantity Distribution:")

display(quantity_distribution)


plt.figure(figsize=(8, 5))

quantity_distribution.plot(
    kind="bar"
)

plt.title("Quantity Distribution")

plt.xlabel("Quantity")

plt.ylabel("Number of Orders")

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 14. MONTHLY SALES TREND
# ------------------------------------------------------------

monthly_sales = (
    df.groupby(
        ["Year", "Month", "Month_Name"]
    )["Total_Sales"]
    .sum()
    .reset_index()
)

monthly_sales = monthly_sales.sort_values(
    ["Year", "Month"]
)

print("\n" + "=" * 70)
print("MONTHLY SALES")
print("=" * 70)

display(monthly_sales)


plt.figure(figsize=(14, 6))

plt.plot(
    monthly_sales["Month_Name"],
    monthly_sales["Total_Sales"],
    marker="o"
)

plt.title("Monthly Sales Trend")

plt.xlabel("Month")

plt.ylabel("Total Sales")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 15. SALES BY AGE GROUP
# ------------------------------------------------------------

age_group_sales = (
    df.groupby(
        "Age_Group",
        observed=True
    )["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("\n" + "=" * 70)
print("SALES BY AGE GROUP")
print("=" * 70)

display(age_group_sales)


plt.figure(figsize=(8, 5))

age_group_sales.plot(
    kind="bar"
)

plt.title("Total Sales by Age Group")

plt.xlabel("Age Group")

plt.ylabel("Total Sales")

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 16. TOP CUSTOMERS
# ------------------------------------------------------------

top_customers = (
    df.groupby(
        ["Customer_ID", "Customer_Name"]
    )["Total_Sales"]
    .sum()
    .sort_values(
        ascending=False
    )
    .head(10)
)

print("\n" + "=" * 70)
print("TOP 10 CUSTOMERS")
print("=" * 70)

display(top_customers)


# ------------------------------------------------------------
# 17. PRODUCT QUANTITY
# ------------------------------------------------------------

product_quantity = (
    df.groupby("Product")["Quantity"]
    .sum()
    .sort_values(ascending=False)
)

print("\n" + "=" * 70)
print("QUANTITY SOLD BY PRODUCT")
print("=" * 70)

display(product_quantity)


plt.figure(figsize=(10, 6))

product_quantity.plot(
    kind="bar"
)

plt.title("Quantity Sold by Product")

plt.xlabel("Product")

plt.ylabel("Quantity")

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 18. CORRELATION ANALYSIS
# ------------------------------------------------------------

correlation = df[
    [
        "Age",
        "Quantity",
        "Unit_Price",
        "Total_Sales"
    ]
].corr()

print("\n" + "=" * 70)
print("CORRELATION MATRIX")
print("=" * 70)

display(correlation)


plt.figure(figsize=(8, 6))

plt.imshow(
    correlation,
    cmap="coolwarm",
    interpolation="nearest"
)

plt.colorbar()

plt.xticks(
    range(len(correlation.columns)),
    correlation.columns,
    rotation=45
)

plt.yticks(
    range(len(correlation.columns)),
    correlation.columns
)

plt.title("Correlation Matrix")

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 19. QUANTITY VS SALES
# ------------------------------------------------------------

plt.figure(figsize=(9, 6))

plt.scatter(
    df["Quantity"],
    df["Total_Sales"],
    alpha=0.5
)

plt.title(
    "Quantity vs Total Sales"
)

plt.xlabel("Quantity")

plt.ylabel("Total Sales")

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 20. UNIT PRICE VS SALES
# ------------------------------------------------------------

plt.figure(figsize=(9, 6))

plt.scatter(
    df["Unit_Price"],
    df["Total_Sales"],
    alpha=0.5
)

plt.title(
    "Unit Price vs Total Sales"
)

plt.xlabel("Unit Price")

plt.ylabel("Total Sales")

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 21. AGE VS SALES
# ------------------------------------------------------------

plt.figure(figsize=(9, 6))

plt.scatter(
    df["Age"],
    df["Total_Sales"],
    alpha=0.5
)

plt.title(
    "Age vs Total Sales"
)

plt.xlabel("Age")

plt.ylabel("Total Sales")

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 22. SAVE EDA SUMMARY
# ------------------------------------------------------------

eda_summary = pd.DataFrame({

    "Metric": [
        "Total Sales",
        "Total Orders",
        "Unique Customers",
        "Total Quantity",
        "Average Sales",
        "Minimum Sale",
        "Maximum Sale",
        "Average Customer Age"
    ],

    "Value": [
        total_sales,
        total_orders,
        unique_customers,
        total_quantity,
        average_sales,
        minimum_sales,
        maximum_sales,
        average_age
    ]
})

eda_summary.to_excel(
    "EDA_Summary.xlsx",
    index=False
)

print("\nEDA completed successfully!")
```
